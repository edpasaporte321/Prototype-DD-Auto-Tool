```python
import pandas as pd
import tkinter as tk
from tkinter import filedialog
import xml.etree.ElementTree as ET
import pathlib
from pathlib import Path
import zipfile
import re

def loadDocx():
    docxPath = filedialog.askopenfilename(
        title="Select a DOCX file",
        filetypes=[("Word files", "*.docx"), ("All files", "*.*")]
    )
    #docxPath = Path(filePath)
    with zipfile.ZipFile(docxPath, 'r') as docx:
        settingsXML = docx.read('word/settings.xml')

    rawXML = settingsXML.decode('utf-8')
    #outputText.delete("1.0", tk.END) 
    #outputText.insert(tk.END, type(rawXML)) - for debug purposes
    return rawXML, root

def loadExcel():
    file_path = filedialog.askopenfilename(
        title="Select an Excel file",
        filetypes=[("Excel files", "*.xlsx *.xls"), ("All files", "*.*")]
    )
    if not file_path:
        return None

    df = pd.read_excel(file_path)

    required_columns = ["Variable Name", "Prompt"]
    missing_columns = [col for col in required_columns if col not in df.columns]

    if missing_columns:
        statusVar.set(f"Missing columns: {', '.join(missing_columns)}")
        outputText.delete("1.0", tk.END)
        outputText.insert(tk.END, f"Missing columns: {', '.join(missing_columns)}\n")
        return

    selected_columns = df[required_columns]

    outputText.delete("1.0", tk.END)
    outputText.insert(tk.END, selected_columns.head().to_string(index=False))
    statusVar.set(f"Loaded: {file_path}")

def extractText(rawXML):
    pattern = r"<Text>(.*?)</Text>"
    matches = re.findall(pattern, rawXML, re.DOTALL)
    print(matches)  # Debugging: print the matches to the console
    return matches

root = tk.Tk()
root.title("Excel File Loader")
root.geometry("700x400")

statusVar = tk.StringVar(value="No file selected")

buttonFrame = tk.Frame(root)
buttonFrame.pack(pady=10)

excelButton = tk.Button(buttonFrame, text="Choose Excel File", command=loadExcel)
excelButton.pack(side="left", padx=(0, 10))

docxButton = tk.Button(buttonFrame, text="Choose DOCX File", command=loadDocx)
docxButton.pack(side="left")

extractTextButton = tk.Button(buttonFrame, text="Extract Text from DOCX", command=extractText)
extractTextButton.pack(side="left", padx=(10, 0))

statusLabel = tk.Label(root, textvariable=statusVar, anchor="w")
statusLabel.pack(fill="x", padx=10)

outputText = tk.Text(root, wrap="word", height=20)
outputText.pack(fill="both", expand=True, padx=10, pady=(0, 10))

root.mainloop()
```
