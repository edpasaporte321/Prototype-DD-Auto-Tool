```python
import pandas as pd
import tkinter as tk
from tkinter import filedialog
import xml.etree.ElementTree as ET
import pathlib
from pathlib import Path
import zipfile
import re

def loadDocx():
    docxPath = filedialog.askopenfilename(
        title="Select a DOCX file",
        filetypes=[("Word files", "*.docx"), ("All files", "*.*")]
    )

    if docxPath == "":
        return None
    
    with zipfile.ZipFile(docxPath, 'r') as docx:
        settingsXML = docx.read('word/settings.xml')

    rawXML = settingsXML.decode('utf-8')

    patternSearch = r"(&lt;Text&gt;&amp;lt;p&amp;gt;)(.*?)(?=&amp;lt;/p&amp;gt;&lt;/Text&gt;)"
    matches = re.findall(patternSearch, rawXML, re.DOTALL)

    outputText.delete("1.0", tk.END)
    for match in matches:
        outputText.insert(tk.END, match[1] + "\n")
    #for match in matches:
        #print(match, "test")
    #outputText.delete("1.0", tk.END) 
    #outputText.insert(tk.END, matches)  # for debugging the matches in the output text widget
    return rawXML, root, matches

def loadExcel():
    filePath = filedialog.askopenfilename(
        title="Select an Excel file",
        filetypes=[("Excel files", "*.xlsx *.xls"), ("All files", "*.*")]
    )
    if not filePath:
        return None

    df = pd.read_excel(filePath)

    requiredColumns = ["Variable Name", "Prompt"]
    missingColumns = [col for col in requiredColumns if col not in df.columns]

    if missingColumns:
        statusVar.set(f"Missing columns: {', '.join(missingColumns)}")
        outputText.delete("1.0", tk.END)
        outputText.insert(tk.END, f"Missing columns: {', '.join(missingColumns)}\n")
        return

    selected_columns = df[requiredColumns]

    outputText.delete("1.0", tk.END)
    outputText.insert(tk.END, selected_columns.head().to_string(index=False))
    statusVar.set(f"Loaded: {filePath}")


rawXML = ""
root = tk.Tk()
root.title("Excel File Loader")
root.geometry("700x1200")

statusVar = tk.StringVar(value="No file selected")

buttonFrame = tk.Frame(root)
buttonFrame.pack(pady=10)

excelButton = tk.Button(buttonFrame, text="Choose Excel File", command=loadExcel)
excelButton.pack(side="left", padx=(0, 10))

docxButton = tk.Button(buttonFrame, text="Choose DOCX File", command=loadDocx)
docxButton.pack(side="left")

#extractTextButton = tk.Button(buttonFrame, text="Extract Text from DOCX", command=lambda: extractText(rawXML))
#extractTextButton.pack(side="left", padx=(10, 0))

statusLabel = tk.Label(root, textvariable=statusVar, anchor="w")
statusLabel.pack(fill="x", padx=10)

outputText = tk.Text(root, wrap="word", height=20)
outputText.pack(fill="both", expand=True, padx=10, pady=(0, 10))

root.mainloop()
```
