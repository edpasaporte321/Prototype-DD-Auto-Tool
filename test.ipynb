```python
import tkinter as tk
from tkinter import filedialog
import pandas as pd


def load_excel():
    file_path = filedialog.askopenfilename(
        title="Select an Excel file",
        filetypes=[("Excel files", "*.xlsx;*.xls"), ("All files", "*.*")]
    )

    if not file_path:
        return

    df = pd.read_excel(file_path)

    output_text.delete("1.0", tk.END)
    output_text.insert(tk.END, df.head().to_string(index=False))
    status_var.set(f"Loaded: {file_path}")


root = tk.Tk()
root.title("Excel File Loader")
root.geometry("700x400")

status_var = tk.StringVar(value="No file selected")

button = tk.Button(root, text="Choose Excel File", command=load_excel)
button.pack(pady=10)

status_label = tk.Label(root, textvariable=status_var, anchor="w")
status_label.pack(fill="x", padx=10)

output_text = tk.Text(root, wrap="word", height=20)
output_text.pack(fill="both", expand=True, padx=10, pady=(0, 10))

root.mainloop()
```

```python
import pandas as pd
import tkinter as tk
from tkinter import filedialog
import xml.etree.ElementTree as ET
import pathlib
from pathlib import Path
import zipfile


def loadDocx():
    docxPath = filedialog.askopenfilename(
        title="Select a DOCX file",
        filetypes=[("Word files", "*.docx"), ("All files", "*.*")]
    )
    #docxPath = Path(filePath)
    with zipfile.ZipFile(docxPath, 'r') as docx:
        settingsXML = docx.read('word/settings.xml')

    rawXML = settingsXML.decode('utf-8')
    return rawXML, root



def loadExcel():
    file_path = filedialog.askopenfilename(
        title="Select an Excel file",
        filetypes=[("Excel files", "*.xlsx *.xls"), ("All files", "*.*")]
    )
    if not file_path:
        return None

    df = pd.read_excel(file_path)

    required_columns = ["Variable Name", "Prompt"]
    missing_columns = [col for col in required_columns if col not in df.columns]

    if missing_columns:
        statusVar.set(f"Missing columns: {', '.join(missing_columns)}")
        outputText.delete("1.0", tk.END)
        outputText.insert(tk.END, f"Missing columns: {', '.join(missing_columns)}\n")
        return

    selected_columns = df[required_columns]

    outputText.delete("1.0", tk.END)
    outputText.insert(tk.END, selected_columns.head().to_string(index=False))
    statusVar.set(f"Loaded: {file_path}")


root = tk.Tk()
root.title("Excel File Loader")
root.geometry("700x400")

statusVar = tk.StringVar(value="No file selected")

button_frame = tk.Frame(root)
button_frame.pack(pady=10)

excel_button = tk.Button(button_frame, text="Choose Excel File", command=loadExcel)
excel_button.pack(side="left", padx=(0, 10))

docx_button = tk.Button(button_frame, text="Choose DOCX File", command=loadDocx)
docx_button.pack(side="left")

statusLabel = tk.Label(root, textvariable=statusVar, anchor="w")
statusLabel.pack(fill="x", padx=10)

outputText = tk.Text(root, wrap="word", height=20)
outputText.pack(fill="both", expand=True, padx=10, pady=(0, 10))

root.mainloop()
```
